```python
df.columns=(
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(r"[^\w]+", "_", regex=True)
    .str.strip( "_")

)
df.columns
```

```python
df.drop_duplicates()
print("Dataset shape after removing duplicates:", df.shape)
```

```python
if "order_date" in df.columns:
  df["order_date"] =  pd.to_datetime(
      df["order_date"],
      errors="coerce"
  )

  if" shape_date" in df.columns:
    df["shape_date"] = pd.to_datetime(
        df["ship_date"],
        errors="coerce"
    )
df[["order_date", "ship_date"]].head()
```

```python
numeric_columns=[
    "sales",
    "profit",
    "Quantity",
    "Discount"
]
for col in numeric_columns:
 if col in df.columns:
    df[col]=pd.to_numeric(
        df[col],
        errors="coerce"
)

df.info()
```

```python
required_columns = [
    col for col in ["sales", "profit", "order_date"]
    if col in df.columns
]

df = df.dropna(subset=required_columns)

print("Rows after cleaning:", len(df))
```

```python
df["profit_margin"] = np.where(
    df["sales"] !=0,
    ( df["profit"] / df["sales"]) * 100,
     np.nan
)

df["order_month"] = df["order_date"].dt.to_period("M")

df["order year"] = df["order_date"] .dt.year

df[[ "sales", "profit", "profit_margin", "order_month"]]. head()
```

```python
# Calculate total sales
total_sales = df["sales"].sum()

# Calculate total profit
total_profit = df["profit"].sum()

# Calculate total orders
total_orders = df["order_id"].nunique()

# Calculate total customers
total_customers = df["customer_id"].nunique()

# Calculate total quantity sold
total_quantity = df["quantity"].sum()

# Calculate overall profit margin
if total_sales != 0:
    overall_profit_margin = (total_profit / total_sales) * 100
else:
    overall_profit_margin = np.nan

# Display business KPIs
print("Total Sales:", round(total_sales, 2))
print("Total Profit:", round(total_profit, 2))
print("Total Orders:", total_orders)
print("Total Customers:", total_customers)
print("Total Quantity Sold:", total_quantity)
print("Overall Profit Margin:", round(overall_profit_margin, 2), "%")
```

```python
category_sales = (
    df.groupby("category")
    .agg(
        total_sales=("sales", "sum"),
        total_profit= ("profit", "sum"),

    )
    .sort_values("total_sales", ascending= False)

)
category_sales
```

```python
category_sales["total_sales"].plot(
    kind= "bar",
    figsize= (8,5)
)

plt.title( "total sales by category")
plt.xlabel("category")
plt.ylabel("sales")
plt.xticks(rotation=0)
plt.show()
```

```python
subcategory_analysis = (
    df.groupby("sub_category")
    .agg(
        total_sales=("sales", "sum"),
        total_profit=("profit", "sum")
    )
    .sort_values("total_profit", ascending=False)
)

subcategory_analysis
```

```python
top_products = (
    df.groupby("product_name")["sales"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_products
```

```python
top_products.sort_values().plot(
    kind="barh",
    figsize=(10, 6)
)

plt.title("Top 10 Products by Sales")
plt.xlabel("Total Sales")
plt.ylabel("Product")
plt.show()
```

```python
region_analysis = (
    df.groupby("region")
    .agg(
        total_sales=("sales", "sum"),
        total_profit=("profit", "sum")
    )
    .sort_values("total_sales", ascending=False)
)

region_analysis
```

```python
segment_analysis = (
    df.groupby("segment")
    .agg(
        total_sales=("sales", "sum"),
        total_profit=("profit", "sum"),
        average_sales=("sales", "mean")
    )
    .sort_values("total_sales", ascending=False)
)

segment_analysis
```

```python
top_customers = (
    df.groupby("customer_name")
    .agg(
        total_sales=("sales", "sum"),
        total_profit=("profit", "sum")
    )
    .sort_values("total_sales", ascending=False)
    .head(10)
)

top_customers
```

```python
monthly_sales = (
    df.groupby("order_month")["sales"]
    .sum()
)

monthly_sales
```

```python
monthly_sales.index = monthly_sales.index.astype(str)

monthly_sales.plot(
    figsize=(12, 5),
    marker="o"
)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.show()
```

```python
discount_analysis = (
    df.groupby("discount")
    .agg(
        total_sales=("sales", "sum"),
        total_profit=("profit", "sum"),
        average_profit=("profit", "mean")
    )
    .sort_index()
)

discount_analysis
```

```python
loss_products = (
    df.groupby("product_name")
    .agg(
        total_sales=("sales", "sum"),
        total_profit=("profit", "sum")
    )
)

loss_products = loss_products[
    loss_products["total_profit"] < 0
].sort_values("total_profit")

loss_products.head(10)
```

```python
numeric_data = df.select_dtypes(include="number")

correlation = numeric_data.corr()

plt.figure(figsize=(10, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Between Numerical Variables")
plt.show()
```

```python
best_category = category_sales["total_sales"].idxmax()
best_region = region_analysis["total_sales"].idxmax()
best_segment = segment_analysis["total_sales"].idxmax()
best_profit_category = category_sales["total_profit"].idxmax()

print("BUSINESS FINDINGS")
print("-----------------")

print(f"Highest sales category: {best_category}")
print(f"Highest sales region: {best_region}")
print(f"Highest sales customer segment: {best_segment}")
print(f"Most profitable category: {best_profit_category}")

print(f"Total sales: {total_sales:,.2f}")
print(f"Total profit: {total_profit:,.2f}")
print(f"Profit margin: {overall_profit_margin:.2f}%")
```
